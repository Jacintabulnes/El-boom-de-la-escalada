```python
import pandas as pd

df = pd.read_csv("/pisa_lectura_limpia.csv")

df.head()
```

```python
chile = df[df["pais_economia"] == "Chile"]

chile
```

```python
chile_2022 = chile[chile["anio"] == 2022].iloc[0]
chile_2025 = chile[chile["anio"] == 2025].iloc[0]

cambio_puntaje = chile_2025["puntaje_lectura"] - chile_2022["puntaje_lectura"]
cambio_bajo_nivel_2 = chile_2025["bajo_nivel_2_pct"] - chile_2022["bajo_nivel_2_pct"]
cambio_nivel_5 = chile_2025["nivel_5_o_superior_pct"] - chile_2022["nivel_5_o_superior_pct"]

print("Cambio en puntaje de lectura:", round(cambio_puntaje, 2))
print("Cambio bajo Nivel 2:", round(cambio_bajo_nivel_2, 2), "puntos porcentuales")
print("Cambio Nivel 5 o superior:", round(cambio_nivel_5, 2), "puntos porcentuales")
```

```python
pisa_2025 = df[df["anio"] == 2025]

comparacion_2025 = pisa_2025[
    ["pais_economia", "puntaje_lectura", "bajo_nivel_2_pct", "nivel_5_o_superior_pct"]
].sort_values("puntaje_lectura", ascending=False)

comparacion_2025
```
