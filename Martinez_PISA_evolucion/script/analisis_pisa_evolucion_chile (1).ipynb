```python
mayor_puntaje = df.loc[df["puntaje_lectura"].idxmax()]
menor_puntaje = df.loc[df["puntaje_lectura"].idxmin()]

print("Ciclo con mayor puntaje:")
print(mayor_puntaje[["anio", "puntaje_lectura"]])

print("\nCiclo con menor puntaje:")
print(menor_puntaje[["anio", "puntaje_lectura"]])
```
